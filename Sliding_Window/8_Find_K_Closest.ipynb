```python
class Solution:
    def findClosestElements(self, arr, k, x):

        # Binary Search
        left = 0
        right = len(arr) - 1

        while left <= right:
            mid = (left + right) // 2

            if arr[mid] < x:
                left = mid + 1
            else:
                right = mid - 1

        # left is insertion position
        r = left
        l = left - 1

        # Expand window
        while k > 0:

            if l < 0:
                r += 1

            elif r >= len(arr):
                l -= 1

            elif abs(arr[l] - x) <= abs(arr[r] - x):
                l -= 1

            else:
                r += 1

            k -= 1

        return arr[l + 1:r]
```
