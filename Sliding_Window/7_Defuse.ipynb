```python
class Solution:
    def decrypt(self, code, k):
        n = len(code)
        ans = [0] * n

        for i in range(n):
            if k > 0:
                for j in range(1, k + 1):
                    ans[i] += code[(i + j) % n]

            elif k < 0:
                for j in range(1, abs(k) + 1):
                    ans[i] += code[(i - j) % n]

        return ans
```
