```python
# Space - O(1)
class Solution:
    def recoverTree(self, root):
        first = second = prev = None

        def inorder(node):
            nonlocal first, second, prev

            if not node:
                return

            inorder(node.left)

            # Inorder should be sorted.
            # If prev > current, we found an inversion.
            if prev is not None:
                if prev.val > node.val:
                    if first is None:
                        first = prev
                    second = node

            prev = node

            inorder(node.right)

        inorder(root)

        first.val, second.val = second.val, first.val
        
# Example:
# Inorder = [1, 5, 3, 4, 2, 6]
#
# Compare adjacent nodes:
# 1 < 5   -> OK
# 5 > 3   -> First inversion
#            first = 5
#            second = 3
# 3 < 4   -> OK
# 4 > 2   -> Second inversion
#            Update second = 2
# 2 < 6   -> OK
#
# Swap first (5) and second (2) to restore the BST.
```
