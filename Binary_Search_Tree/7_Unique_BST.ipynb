```python
class Solution:
    def numTrees(self, n: int) -> int:
        dp = [0] * (n + 1)
        dp[0] = dp[1] = 1

        for nodes in range(2, n + 1):
            for root in range(1, nodes + 1):
                left = root - 1
                right = nodes - root
                dp[nodes] += dp[left] * dp[right]

        return dp[n]

# dp[i] stores the number of unique BSTs that can be formed using i nodes.
#
# For each possible root j (1 to i):
# - Left subtree has (j - 1) nodes.
# - Right subtree has (i - j) nodes.
#
# Number of BSTs with j as the root:
#     dp[j - 1] * dp[i - j]
#
# Sum this for every possible root:
#     dp[i] += dp[j - 1] * dp[i - j]
```
