```python
class Solution:
    def kthSmallest(self, root, k):
        ans = None

        def inorder(node):
            nonlocal k, ans

            if not node or ans is not None:
                return

            inorder(node.left)

            k -= 1
            if k == 0:
                ans = node.val
                return

            inorder(node.right)

        inorder(root)
        return ans
```
