# Pure Recursion

```python
class Solution:
    def stoneGameII(self, piles: List[int]) -> int:

        def helper(i, Alice_turn, M):

            if i >= len(piles):
                return 0

            if Alice_turn:
                best = 0

                for X in range(1, 2 * M + 1):
                    if i + X > len(piles):
                        break

                    current = sum(piles[i:i + X])

                    future = helper(
                        i + X,
                        False,
                        max(M, X)
                    )

                    best = max(best, current + future)

                return best

            else:
                best = float("inf")

                for X in range(1, 2 * M + 1):
                    if i + X > len(piles):
                        break

                    future = helper(
                        i + X,
                        True,
                        max(M, X)
                    )

                    best = min(best, future)

                return best

        return helper(0, True, 1)
```

# Memo

```python
class Solution:
    def stoneGameII(self, piles: List[int]) -> int:

        memo = {}

        def helper(i, Alice_turn, M):

            if i >= len(piles):
                return 0

            if (i, Alice_turn, M) in memo:
                return memo[(i, Alice_turn, M)]

            if Alice_turn:
                best = 0

                for X in range(1, 2 * M + 1):
                    if i + X > len(piles):
                        break

                    current = sum(piles[i:i + X])

                    future = helper(
                        i + X,
                        False,
                        max(M, X)
                    )

                    best = max(best, current + future)

            else:
                best = float("inf")

                for X in range(1, 2 * M + 1):
                    if i + X > len(piles):
                        break

                    future = helper(
                        i + X,
                        True,
                        max(M, X)
                    )

                    best = min(best, future)

            memo[(i, Alice_turn, M)] = best
            return best

        return helper(0, True, 1)
```
