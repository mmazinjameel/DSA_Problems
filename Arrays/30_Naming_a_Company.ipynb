```python
class Solution:
    def distinctNames(self, ideas: List[str]) -> int:
        
        # Group Suffixes by first letter
        groups = [set() for _ in range(26)]

        for idea in ideas:
            first = ord(idea[0]) - ord('a')
            suffix = idea[1:]
            groups[first].add(suffix)

        ans = 0

        # Compare every pair of first-letter groups
        for i in range(26):
            for j in range(i + 1, 26):
                # Common suffixes between the two groups
                common = groups[i] & groups[j]

                # Suffixes unique to each group
                unique_i = len(groups[i]) - len(common)
                unique_j = len(groups[j]) - len(common)

                # Each valid pair gives two company names
                ans += unique_i * unique_j * 2

        return ans
```
