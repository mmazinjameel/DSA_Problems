```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def sumNumbers(self, root: Optional[TreeNode]) -> int:
        
        def helper(node, curr_num):
            if not node:
                return 0
            
            curr_num = curr_num * 10 + node.val  # Update the number

            # If it's a leaf node, return the formed number
            if not node.left and not node.right:
                return curr_num

            # Sum numbers from left and right subtrees
            return helper(node.left, curr_num) + helper(node.right, curr_num)

        return helper(root, 0)
```
