```python
class Solution:
    def countPairs(self, root, distance):
        self.ans = 0

        def dfs(node):
            if not node:
                return []

            # Leaf node
            if not node.left and not node.right:
                return [1]

            left = dfs(node.left)
            right = dfs(node.right)

            # Count good pairs
            for l in left:
                for r in right:
                    if l + r <= distance:
                        self.ans += 1

            # Return distances to parent
            res = []

            for d in left:
                if d + 1 <= distance:
                    res.append(d + 1)

            for d in right:
                if d + 1 <= distance:
                    res.append(d + 1)

            return res

        dfs(root)
        return self.ans
```
