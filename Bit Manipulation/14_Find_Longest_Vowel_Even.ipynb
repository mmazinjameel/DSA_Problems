```python
class Solution:
    def findTheLongestSubstring(self, s: str) -> int:


        first = {0: -1}
        mask = 0
        ans = 0

        bits = {
            'a': 0,
            'e': 1,
            'i': 2,
            'o': 3,
            'u': 4
        }

        for i, ch in enumerate(s):
            if ch in bits:
                mask ^= (1 << bits[ch])

            if mask in first:
                ans = max(ans, i - first[mask])
            else:
                first[mask] = i

        return ans

        # Use a 5-bit mask to track whether each vowel's count is even (0) or odd (1).
        # If the same mask appears again, the substring between the two indices has even counts of all vowels.
```
