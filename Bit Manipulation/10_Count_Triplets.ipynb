```python
class Solution:
    def countTriplets(self, arr: List[int]) -> int:
        n = len(arr)
        ans = 0

        for i in range(n):
            xor = 0
            for k in range(i, n):
                xor ^= arr[k]

                if xor == 0:
                    ans += k - i

        return ans

        # Key idea:
        # 1. Fix the starting index i.
        # 2. Keep extending the ending index k while maintaining the running XOR.
        # 3. Whenever XOR of arr[i...k] becomes 0, every split point j
        #    (where i < j <= k) satisfies:
        #       XOR(arr[i...j-1]) == XOR(arr[j...k])
        # 4. The number of valid split points is exactly (k - i),
        #    so instead of checking every j, simply do:
        #       ans += (k - i)
```
