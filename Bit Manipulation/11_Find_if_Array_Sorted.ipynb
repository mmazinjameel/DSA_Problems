```python
class Solution:
    def canSortArray(self, nums: List[int]) -> bool:
        n = len(nums)
        prev_max = float("-inf")
        i = 0

        while i < n:
            bits = nums[i].bit_count()
            curr_min = nums[i]
            curr_max = nums[i]

            # Process the current contiguous segment
            while i < n and nums[i].bit_count() == bits:
                curr_min = min(curr_min, nums[i])
                curr_max = max(curr_max, nums[i])
                i += 1

            # Smallest value in this segment cannot be smaller than
            # the largest value from any previous segment.
            if curr_min < prev_max:
                return False

            prev_max = max(prev_max, curr_max)

        return True


# Logic:
#
# 1. Adjacent swaps are only allowed when two numbers have the same popcount.
# 2. Therefore, elements can never cross a boundary where the popcount changes.
# 3. This partitions the array into contiguous segments with the same popcount.
# 4. Within each segment, any permutation is possible (adjacent swaps can realize
#    any ordering), so only the segment's minimum and maximum matter.
# 5. Scan each segment once:
#       - Compute its minimum and maximum.
#       - Keep track of the maximum value seen in all previous segments.
# 6. If the current segment's minimum is smaller than the previous maximum,
#    that value would need to cross a segment boundary to achieve sorted order,
#    which is impossible. Return False.
# 7. Otherwise, update prev_max and continue.
# 8. If all segments satisfy the condition, return True.
#
# Time: O(n)
# Space: O(1)
```
