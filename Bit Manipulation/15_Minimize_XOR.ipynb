```python
class Solution:
    def minimizeXor(self, num1: int, num2: int) -> int:
        need = bin(num2).count("1")
        ans = 0

        # Take set bits from num1 (left to right)
        for i in range(31, -1, -1):
            if (num1 & (1 << i)) and need:
                ans |= (1 << i)
                need -= 1

        # Add remaining set bits from right to left
        for i in range(32):
            if need == 0:
                break
            if (ans & (1 << i)) == 0:
                ans |= (1 << i)
                need -= 1

        return ans

# Match the highest set bits of num1 first to minimize XOR.
# If more set bits are needed, add them from the lowest bits.
```
