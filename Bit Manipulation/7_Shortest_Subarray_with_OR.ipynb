```python
class Solution:
    def minimumSubarrayLength(self, nums: List[int], k: int) -> int:

        n = len(nums)
        cnt = [0] * 32
        left = 0
        cur_or = 0
        ans = float("inf")

        for right in range(n):
            # Add nums[right]
            for b in range(32):
                if nums[right] & (1 << b):
                    cnt[b] += 1
                    cur_or |= (1 << b)

            # Shrink window while OR >= k
            while left <= right and cur_or >= k:
                ans = min(ans, right - left + 1)

                # Remove nums[left]
                for b in range(32):
                    if nums[left] & (1 << b):
                        cnt[b] -= 1
                        if cnt[b] == 0:
                            cur_or &= ~(1 << b)

                left += 1

        return ans if ans != float("inf") else -1
        
# Every time you add or remove a number, update the 32 bit counts.
```
