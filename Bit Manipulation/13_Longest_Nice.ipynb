```python
class Solution:
    def longestNiceSubarray(self, nums: List[int]) -> int:
        left = 0
        mask = 0
        ans = 0

        for right in range(len(nums)):
            while mask & nums[right]:
                mask ^= nums[left]
                left += 1

            mask |= nums[right]
            ans = max(ans, right - left + 1)

        return ans
        
# Instead of checking every pair of numbers, keep one variable called mask.
# mask remembers which bits are already occupied in the current window.
# If the result is 0, none of its bits are occupied → add it.
# If the result is not 0, some bit is already occupied → remove numbers from the left until there's no conflict.
```
