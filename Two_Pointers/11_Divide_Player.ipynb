```python
class Solution:
    def dividePlayers(self, skill):
        skill.sort()

        total = sum(skill)
        teams = len(skill) // 2
        team_sum = total // teams

        left = 0
        right = len(skill) - 1
        chemistry = 0

        while left < right:
            if skill[left] + skill[right] != team_sum:
                return -1

            chemistry += skill[left] * skill[right]

            left += 1
            right -= 1

        return chemistry
```
