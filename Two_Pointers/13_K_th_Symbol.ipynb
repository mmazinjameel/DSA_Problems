```python
class Solution:
    def kthGrammar(self, n: int, k: int) -> int:
        # Base case
        if n == 1:
            return 0

        # Find the position in the previous row
        parent = (k + 1) // 2

        # Get the value from the previous row
        value = self.kthGrammar(n - 1, parent)

        # If k is even, we take the second character,
        # which is the opposite of the parent.
        if k % 2 == 0:
            return 1 - value

        # If k is odd, we take the first character,
        # which is the same as the parent.
        return value
        
# Second character is always the opposite of its parent.

# Example: parent = 0 → 01, so second character = 1.
```
