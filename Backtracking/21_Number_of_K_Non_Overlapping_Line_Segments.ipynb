# Recursion

```python
class Solution:
    def numberOfSets(self, n: int, k: int) -> int:
        def solve(i, k, started):

            if k == 0:
                return 1

            if i == n:
                return 0

            # Skip this point
            ans = solve(i + 1, k, started)

            if started:
                # End the current segment at i
                ans += solve(i, k - 1, False)

            else:
                # Start a segment at i
                ans += solve(i + 1, k, True)

            return ans

        return solve(0, k, False)
```

# Memoization

```python
class Solution:
    def numberOfSets(self, n: int, k: int) -> int:
        MOD = 10**9 + 7
        memo = {}

        def solve(i, k, started):
            if k == 0:
                return 1

            if i == n:
                return 0

            key = (i, k, started)

            if key in memo:
                return memo[key]

            # Skip this point
            ans = solve(i + 1, k, started)

            if started:
                # End the current segment at i
                ans += solve(i, k - 1, False)

            else:
                # Start a segment at i
                ans += solve(i + 1, k, True)

            memo[key] = ans % MOD
            return memo[key]

        return solve(0, k, False)
```
